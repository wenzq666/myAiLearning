## 创建数组

```python
arr1 = np.arange(15).reshape(3,5)
print(arr1)
```

## 常用属性

### array()  python列表转ndarray

```python
my_list = [7,8,9,4,5,6,1,23]

# 把python中的列表转numpy的ndarray
arr2 = np.array(my_list)
print(arr2)
print(type(arr2))
```

### arange 创建ndarray对象

```python
# arange(起始, 结束, 步长, dtype = 类型)
arr3 = np.arange(0, 10, 2,dtype=int)
print(arr3)
print(type(arr3))
print(arr3.dtype)
```

### 随机数生成 ndarray

```python
# random.rand()  生成 [0,1)之间 左闭右开的小数矩阵
arr4 = np.random.rand(3, 5)
print(arr4)
```

```python
# random.randint(起始, 结束, size=(行,列)) 生成指定范围 左闭右开 的随机整数矩阵
arr5 = np.random.randint(-5, 10, size=(3,4))
print(arr5)
```

```python
# random.uniform(起始, 结束, size=(行,列)) 生成指定范围 左闭右开 的随机小数矩阵
arr6 = np.random.uniform(-5, 10, size=(3,4))
print(arr6)
```

### numpy类型转换 astype()

```python
arr7 = np.arange(0, 10, 2)
print(arr7)
arr8 = arr7.astype(np.float64)
print(arr8)
print(arr8.dtype)
```

### 等比/等差数列

```python
# np.logspace 等比数列
# np.logspace(0,10,10) 开始 和 结束是10的幂次方(默认底数是10) 闭区间 元素个数为10
# np.logspace(0,0,10,base=2) 底数为2
a = np.logspace(0,9,10)
print(a)
aa = np.logspace(0,3,4,base=2)
print(aa)

# np.linspace 等差数列
# np.linspace(1,10,10,endpoint=True) 开始,结束,元素个数  闭区间
#                     endpoint(默认为True 右闭区间)
b = np.linspace(1,10,10,endpoint=False)
print(b)
```

# numpy常用函数

### 基本函数

```python
# 创建正态分布标准序列
arr10 = np.random.randn(3,5) # 生成(−∞,+∞)之间的3行5列矩阵
print(arr10)
```

```python
# 基本函数
print(np.ceil(arr10)) # 向上取整
print(np.floor(arr10)) # 向下取整
print(np.rint(arr10)) # 四舍五入
print(np.abs(arr10)) # 绝对值
# 矩阵相乘
#print(np.multiply(arr10, arr10))
print(arr10 * arr10)

# where 类似三元表达式
print(np.where(arr10 > 0, 1, -1)) #arr10的每个元素去做判断
```

```python
# 内置函数
arr11 = np.arange(12).reshape(3,4)
print(arr11)
```

```python
# 求平均值
print(np.mean(arr11))
# 求元素的和
print(np.sum(arr11))
# 按 列 统计和 返回一维数组
print(np.sum(arr11,axis=0))
# 按 行 统计和 返回一维数组
print(np.sum(arr11,axis=1))

print(np.cumsum(arr11))
```

### 去重/排序函数

```python
# 排序定义数组
arr15 = np.array([1,2,1,6,5,3])
print(arr15)
```

### 矩阵乘法

```python
# 定义数组
arr1 = np.array([[1,2,3],[4,5,6]])
arr2 = np.array([[1,2,3],[4,5,6]])
print(arr1)
print(arr2)
```

```python
# 矩阵元素乘法(哈达玛积)  行列数一致(对应坐标的元素做乘法)
print(arr1 * arr2)
print(np.multiply(arr1,arr2))
```

```python
# 矩阵乘法  行列数不一致
arr3 = np.array([[1,2,3],[4,5,6]])
arr4 = np.array([[9,8],[6,5],[3,2]])
print(arr3)
print(arr4)
```

```python
# 矩阵乘法(线代)
# 第一个矩阵的列 = 第二个矩阵的行
# A列 = B行  结果是 A行B列
print(np.dot(arr3,arr4))

print(arr3.dot(arr4))

#语法糖
print(arr3 @ arr4)
```
