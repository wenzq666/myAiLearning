# pandas

```python
import pandas as pd
import numpy as np
```

### pandas两大核心对象
#### series 一列数据     DataFrame 多列数据

```python
# 创建series对象  默认自增索引
s1 = pd.Series([1,2,3,4,5,6])
print(s1)
```

```python
# 创建series对象  自定义索引
s2 = pd.Series([1,2,3,4,5,6],index=['a',11,13,14,15,18])
print(s2)
```

```python
# 使用元组创建Series对象
s3 = pd.Series((2,4,6,8,10))
print(s3)
```

```python
# 使用字典创建Series对象
s4 = pd.Series({'a':2,'b':3,'c':5,'d':8})
print(s4)
```

```python
# 使用numpy创建Series对象
s5 = pd.Series(np.arange(5))
print(s5)
```

```python
s6 = pd.Series(data=[i for i in range(6)],index=[i for i in 'ABCDEF'])
print(s6)
```

```python
# 获取Series对象的索引值
print(s6.index)
```

```python
# 获取Series对象的值
print(s6.values)
```

```python
# 获取Series对象的 索引对应的元素
print(s6['F'])

# 根据索引修改Series对象的元素值
s6['F'] = 100

print(s6)
```

# DataFrame 对象

```python
# 创建dataframe
# 字典 + 列表
# 每个键值对 = 1列数据
data = {
    'name':['迪丽热巴', '古力娜扎', '马儿扎哈', '易烊千玺'],
    'gender':['女', '女', '女', '男'],
    'age':[18, 19, 20, 21]
}

# 封装
df1 = pd.DataFrame(data)

print(df1)
```

```python
# 列表 +元组/列表 创建 df对象
# 每个键值对 = 1行数据  字典默认是索引
data = [
    ('迪丽热巴','女',18),
    ('古力娜扎','女',19),
    ('马儿扎哈','女',20),
    ('易烊千玺','男',21)
]

df2 = pd.DataFrame(data,columns=['name','gender','age'])

df2
```

#### 学生信息处理案例

```python
# 10名同学 5门功课  成绩范围40-100
# 生成10行5列  的 40-100范围的数据
score_df = pd.DataFrame(np.random.randint(40,101,(10,5)))
score_df
```

```python
# 修改DataFarme对象的 列名 和 索引列值
column_names = ['数学','语文','英语','政治','体育']
index_names = ['同学'+str(i) for i in range(score_df.shape[0])]
print(index_names)
# score_df.columns = column_names
# score_df.index=index_names

score_df.rename(
    index={i:index_names[i] for i in range(score_df.shape[0])},
    columns={i:column_names[i] for i in range(score_df.shape[1])},
    inplace=True
)

score_df
```
